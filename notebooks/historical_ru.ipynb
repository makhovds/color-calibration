### Что сохраняет этот вариант

Сохраняются исходные численные решения: сопоставление спектров по индексам, нормировка OD стекол, вещественные RGB байтового масштаба, квантование спектральных эталонов через RGB и 98 строк обучения с осветителем. Они воспроизводят прежний расчет и не являются рекомендацией для новых измерений. Комплексные OD и нестандартная историческая ошибка отмечены явно. Альтернативный расчет — [corrected_ru.ipynb](corrected_ru.ipynb).

## 9. Полиномиальная коррекция внутри каждого микроскопа

Обучаем **зарегистрированные Lab → спектральные эталонные Lab**, отдельно для Meiji и Olympus. Сравниваем степени 1–5 с перекрестными членами и без них. Исторический показатель и стандартное среднее ΔE76 различаются:

* Стандартное среднее: усреднение `sqrt(ΔL² + Δa² + Δb²)` по полям.
* Исторический показатель: `(sqrt(sum(ΔL²)) + sqrt(sum(Δa²)) + sqrt(sum(Δb²))) / N`.

Исторический показатель воспроизводит округленные **0.75 и 0.66**, но его нельзя называть средним ΔE76. [Эксперимент и определения](../docs/ru/calibration.md).

```python
from analysis.calibration import run_experiment
comparison, models, pairs = run_experiment(config, PROFILE)
figures.plot_calibration(config, comparison, PROFILE, LANGUAGE)
reporting.selected_models(comparison, LANGUAGE)
```

Модели используют 98 строк с осветителем. Без перекрестных членов исходная реализация обучает выход только по одноименному входному каналу; это отличается от аддитивной формулы всех каналов в статье. Крупные исторические Lab делают высокие степени численно чувствительными. Рост ошибки обучения сам по себе не доказывает переобучение.

Ниже — **отдельная аппроксимация архивных Lab-таблиц** для сверки округленных результатов статьи. Она не заменяет выполненный выше пересчет спектров/изображений. Последний столбец показывает расхождение новых и сохраненных зарегистрированных Lab; для Olympus точной идентичности нет.
