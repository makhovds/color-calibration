# Microscopy color calibration
## Explicitly corrected workflow

Follow the original data from spectra and images to a gamut comparison and per-microscope polynomial calibration. Run cells in order, or choose **Restart Kernel and Run All Cells**.

**Read first:** [data setup](../docs/en/data.md), [hardware](../docs/en/hardware.md), [sources and preprocessing](../docs/en/preprocessing.md). Download the spectra/reference and image bundles, then set `data_root` in `config.json`. Code is shared with the [Russian notebook](corrected_ru.ipynb); only the explanation and display language change.

This is the official research workflow for [the calibration-slide paper](https://doi.org/10.1007/s11018-022-02014-7) and [the polynomial-calibration paper](https://doi.org/10.1088/1742-6596/2192/1/012010). The inputs are the surviving research files. The published 99.4% gamut coverage cannot be recovered exactly from this available snapshot.

### What is corrected

We interpolate wavelengths when necessary, mask nonpositive transmission, preserve glass OD magnitude, normalize RGB/XYZ scales, calculate spectral Lab directly, exclude the illuminator from fitting, stabilize polynomial inputs, and report standard mean ΔE76 with held-out errors. The no-cross polynomial follows the additive feature definition printed in the paper. [All differences and remaining assumptions](../docs/en/corrections.md) are documented; camera sRGB encoding is still an assumption. This is a new analysis of the same data, not a replacement for the historical result.

## 9. Polynomial correction within each microscope

Fit **registered Lab → spectral-reference Lab**, independently for Meiji and Olympus. Compare degrees 1–5 with and without cross terms. The plotted historical aggregate and standard mean ΔE76 are different quantities:

* Standard mean: average of `sqrt(ΔL² + Δa² + Δb²)` over fields.
* Archived aggregate: `(sqrt(sum(ΔL²)) + sqrt(sum(Δa²)) + sqrt(sum(Δb²))) / N`.

The archived aggregate reproduces the paper's rounded **0.75 and 0.66**, but must not be called mean ΔE76. [Experiment and metric definitions](../docs/en/calibration.md).

```python
from analysis.calibration import run_experiment
comparison, models, pairs = run_experiment(config, PROFILE)
figures.plot_calibration(config, comparison, PROFILE, LANGUAGE)
reporting.selected_models(comparison, LANGUAGE)
```

The corrected fit uses 97 fields, direct spectral Lab and normalized assumed-sRGB inputs. Centering and scaling are learned from the training data. The no-cross model includes powers of all input channels, following the printed additive formula.

The next table compares **training and five-fold held-out errors**. Entire fields are held out; these are not independent acquisition sessions. Large degree-5 errors are informative and remain visible. We still use the paper's degrees 3/4 for the image demonstration, rather than silently changing the experiment.

```python
columns = ["system", "degree", "cross_terms", "mean_delta_e76", "cv_mean_delta_e76"]
reporting.table(comparison[columns], LANGUAGE)
```
